```python
# How do we replace the null values?
# The answer is always that we replace it with  median or mean
# Generally u must  choose median over mean  because mean could get affected by outliers and that could skew the distribution of our data
# And why median because median is robust to outliers
# Common step beginner do is fill all the missing values with the overall median of the entire column(This technically works)
# But this is not the best practice bcoz reviews vary with different  category(eg:footwear rating differ from cloth review rating)
# so filling median of footwear overall will force all category to follow the rating pf footwear which can introduce bias into our datasets 
##### WE WILL IMPUTE THE MISSING VALUES USING THE MEDIAN OF REVIEW RATINGS WITHIN EACH CATEGORY.
# WE WILL FIND MEDIAN REVIEW RATING WITHING EACH CATEGORY ANND WILL REPLACE THAT WITH THE MISSING VALUE. THAT IS ,IF THERE IS A MISSING REVIEW RATING IN THE CLOTHING CATEGORY
# WE WILL REPLACE IT WITH THE MEDIAN REVIEW RATING OF THE CLOTHING ITEM NOT FOOTWEAR ITEM
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Always look at the format of column names like the starting letter uppercase then there's a space so we need 
# to remember this as we have to use it in the code but isnt it difficult to remember the format of every column?
# So we use SNAKE CASING in all columns. its all lowe case letter with underscores
# It is easy to refrence column and the format is consistent so its easy to write in SQL or Python
# eg:customer_id
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#Create a column age group   (feature)
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#Feature2: Create column purchase_frequency_days
# (Ok so right now in the dataset we have a column purchase frequency which tells us how aften a customer purchases 
# in text form like monthly,quaterly,etc and analyzing this data could be little less accurate and harder
# (especially if u want to calculate metrics or compare customersr
# So will convert textual frequency into numerical values(number of days of purchases) which makes calculations easier
#Below  is a Dictionary that maps textual frequency into equivalent number of days 
frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
